```python
df = pd.read_csv('/content/drive/MyDrive/AST17/POWER ACTIVITY /AST17_POWER DAV_SEOUL, SOUTH KOREA (YEAR 2004).csv')
df
```

```python
df.columns = df.columns.str.strip()
df['YEAR 2004'] = pd.to_datetime(df['YEAR 2004'], format='%b')
months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']
plt.figure(figsize=(12, 6))

plt.plot(months, df['T2M'], label='Temperature at 2 Meters (C)', color='blue', marker='o')
for i, val in enumerate(df['T2M']):
    plt.text(months[i], val + 1, f"{val:.2f}", ha='right', va='baseline', fontsize= 9, color = 'blue')
plt.plot(months, df['QV2M'], label='Specific Humidity at 2 Meters (g/kg)', color='purple', marker='x')
for i, val in enumerate(df['QV2M']):
    plt.text(months[i], val + 1.5, f"{val:.2f}", ha='left', va='baseline', fontsize= 9, color = 'purple')
plt.plot(months, df['RH2M'], label='Relative Humidity at 2 Meters (%)', color='green', marker='s')
for i, val in enumerate(df['RH2M']):
    plt.text(months[i], val, f"{val:.2f}", ha='center', va='bottom', fontsize= 9, color = 'green')

plt.title('Monthly Temperature at 2 Meters (C), Specific Humidity at 2 Meters (g/kg), Relative Humidity at 2 Meters (%) for 2004')
plt.xlabel('Months')
plt.ylabel('Values')
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```

# **Explanation:**

***Intepretation***

The graph as shown above, offers a detailed look at Seoul, South Korea's climate in 2004, which shows the monthly changes in its three parameters: temperature, specific humidity, and relative humidity. It shows how these factors interact to shape the city's overall weather patterns.

Temperature (T2M) are evidently shown in the graph, with its peak in July at around 24.2°C, which is the hottest month, and drop to their lowest in January at around -2.8°C, which is the coldest. This pattern or trend usually happens for seasonal climate, with summer (June to August) being warmer and winter (December to February) being colder.

Specific humidity (QV2M) closely mirrors the temperature changes. As temperatures rise, the air holds more moisture, leading to higher specific humidity levels. This is evident in the graph, where both temperature and specific humidity peak in July at around 16.54 g/kg, and hit their lowest points in January at around 2.32 g/kg. Therefore, it suggests that there is a strong connection between temperature and the air's moisture content.

Relative humidity (RH2M), however, tells a more complex side. While it tends to be lower during the warmer months, and higher in the colder ones, with July having the highest RH2M at 89.38% and March the lowest at 71.69%, the pattern is not as straightforward as it is with specific humidity. This suggests that factors like atmospheric pressure and wind also play a role in determining relative humidity levels.

***Correlation between the three parameters***

The graph reveals how these parameters—temperature, specific humidity, and relative humidity—work together.Think of it like this scenario: warmer air is like a bigger sponge, able to soak up and hold more moisture. That is why specific humidity (the actual amount of water vapor in the air) increases as the temperature increases.

Next, relative humidity is a bit trickier. It is more about how much moisture is in the air compared to how much it could hold at that temperature. So, even though warmer air can hold more moisture, it does not always mean it will be more humid. Sometimes, other factors like wind and air pressure come into play, affecting how much moisture is actually present in the air relative to its capacity.

Lastly, specific and relative humidity often move in similar ways, they can sometimes differ, probably because of changes in the overall atmospheric conditions.
